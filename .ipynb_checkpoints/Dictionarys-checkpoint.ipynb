```python
s1='virat.kohli@blr.com,Rohit.sharma@mumbai,kl.rahul@lucknow.com'
name=[]
firstname=[]
secondname=[]
company=[]
a1=s1.split(',')
for i in a1:
     if '.' in i:
        name.append(i.split('.'))
print(name) 
for i in range(len(name)):
    firstname.append(name[i][0])
    l=name[i][1].split('@')
    secondname.append(l[0])
    company.append(l[1])
print("firstnames are:",firstname)
print("firstnames are:",secondname) 
print("firstnames are:",company)
```
