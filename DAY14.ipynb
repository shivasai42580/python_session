# NESTED IF

```python
num=eval(input("enter a number:"))
if(num%2==0):
    if(num>0):
        print("{} is positive number".format(num))
    elif(num<0):
        print("{} is negative number".format(num))
    else:
        print("It is zero")
else:
    if(num%2!=0):
        print("Square of {} is {}".format(num,num*num))
```

```python
num=eval(input("enter a number:"))
if(num%5==0):
    if(num%2==0):
        print("{} is even number".format(num))
    elif(num%2!=0):
        print("{} is odd number".format(num))
else:
    print("The last digit of {} is {}".format(num,num%10))
```

```python
num=eval(input("enter a number:"))
if(num>0):
    if(num*num%5==0):
        print("{} is divisible by 5".format(num*num))
    elif(num>50):
        print("{} is greater than 50".format(num))
    else:
        print("the number is:",num)
else:
    if(num<0):
        if(num>-50 and num<-10):
            print("{} is b/w -10 to -50".format(num))
    else:
        print("The number is Zero")
 
```

```python
#greatest among 3 numbers
num1=eval(input("enter  number1:"))
num2=eval(input("enter  number2:"))
num3=eval(input("enter  number3:"))
if(num1>num2):
    if(num1>num3):
        print("{} is greater than {} and {}".format(num1,num2,num3))
    else:
        print("{} is greater than {} and {}".format(num3,num2,num1))
else:
    if(num2>num3):
        print("{} is greater than {} and {}".format(num2,num1,num3))
    else:
        print("{} is greater than {} and {}".format(num3,num2,num1))
```
