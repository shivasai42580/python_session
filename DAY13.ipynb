```python
#Greatest Among Three Numbers
num1=eval(input("enter number1:"))
num2=eval(input("enter number2:"))
num3=eval(input("enter number3:"))
if(num1>num2 and num1>num3):
    print("{} is greatest among {} and {}".format(num1,num2,num3))
elif(num2>num3):
    print("{} is greatest among {} and {}".format(num2,num1,num3))
elif(num3>num2):
    print("{} is greatest among {} and {}".format(num3,num1,num2))
```

```python
Principal=eval(input("Enter Total amount of loan:"))
Interest_rate=eval(input("Enter Rate of Interest per annum:"))
Tenure=eval(input("Enter Total tenure of loan in years:"))
Interest=(Principal*Interest_rate*Tenure)/100
if(Interest_rate<5):
    print("The Interst rate of ₹{} is fine for {} years for loan amount of ₹{}".format(Interest,Tenure,Principal))
else:
    print("High Interst rate")
```
